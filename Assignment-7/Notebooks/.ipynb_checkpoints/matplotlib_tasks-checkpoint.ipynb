```python
import matplotlib.pyplot as plt
```

```python
# Task 6:

names = ['Ali', 'Aruzhan', 'Dias', 'Amina', 'Nursultan']
scores = [85, 90, 78, 88, 92]

plt.plot(names, scores)
plt.xlabel('Students')
plt.ylabel('Scores')
plt.title('Student Scores Line Plot')

plt.show()
```

```python
# Task 7:

names = ['Ali', 'Aruzhan', 'Dias', 'Amina', 'Nursultan']
scores = [85, 90, 78, 88, 92]

plt.bar(names, scores)
plt.xlabel('Students')
plt.ylabel('Scores')
plt.title('Student Scores Bar Chart')

plt.show()
```
